## 1.1 Bedingte Erwartung

Implementierung und Visualisierung der bedingten Erwartung.

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
import os
sys.path.append(os.path.dirname(os.path.abspath('')))
from martingales import *

# Beispiel: Bedingte Erwartung
np.random.seed(42)
X = np.random.normal(0, 1, 10000)
Y = X + np.random.normal(0, 0.5, 10000)  # Y ist korreliert mit X

# Berechne E[X | Y = y] für verschiedene y-Werte
y_values = np.linspace(-3, 3, 20)
conditional_expectations = []

for y in y_values:
    mask = (np.abs(Y - y) < 0.2)  # Näherung für diskrete Werte
    if np.sum(mask) > 0:
        cond_exp = np.mean(X[mask])
        conditional_expectations.append(cond_exp)
    else:
        conditional_expectations.append(np.nan)

plt.figure(figsize=(10, 6))
plt.plot(y_values, conditional_expectations, 'o-', linewidth=2, markersize=8)
plt.plot([-3, 3], [-3, 3], 'r--', linewidth=2, label='y = x (theoretisch)')
plt.xlabel('Y = y', fontsize=12)
plt.ylabel('E[X | Y = y]', fontsize=12)
plt.title('Bedingte Erwartung E[X | Y]', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
